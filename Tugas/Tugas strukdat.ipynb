```python
#cek palindrome
def cek_palindrome(teks):
    stack = []
    for huruf in teks:
        stack.append(huruf)
    for huruf in teks:
        if huruf != stack.pop():
            return "Bukan Palindrome"
    
    return "Palindrome"
print(cek_palindrome(input("Masukkan kata: ")))
```

```python
stack = []
while True:
    teks = input("Masukkan kata (ketik Y untuk berhenti): ") 
    if teks == "Y":
        break
    elif teks == "UNDO":
        if stack:
            stack.pop()
        else:
            print("Tidak bisa di-undo")
    else:
        stack.append(teks)
    if stack:
        print("Hasil:", stack[-1])
    else:
        print("Kosong")
```

```python
kalimat = input("masukkan Kalimat:")
stack = []

kata = kalimat.split()

for k in kata:
    stack.append(k)

hasil = []
while stack:
    hasil.append(stack.pop())

print(" ".join(hasil))
```

```python
stack = [10, 20, 30, 40, 50]

jumlah = 0
for _ in stack:
    jumlah += 1

print("Jumlah elemen:", jumlah)
```

```python
stack = [10, 20, 30, 40, 50]
temp = []

while stack:
    nilai = stack.pop()
    if nilai != 30:
        temp.append(nilai)

while temp:
    stack.append(temp.pop())

print(stack)
```

```python
maks = 5
parkiran = []

while True:
    print("1. Mobil Masuk")
    print("2. Mobil Keluar")
    print("3. Lihat Parkiran")
    print("4. Keluar")
    
    pilih = input("Pilih: ")
    
    if pilih == "1":
        if len(parkiran) < maks:
            mobil = input("Nomor mobil: ")
            parkiran.append(mobil)
        else:
            print("Parkiran Penuh!")
    
    elif pilih == "2":
        if parkiran:
            keluar = parkiran.pop()
            print("Mobil keluar:", keluar)
        else:
            print("Parkiran kosong!")
    
    elif pilih == "3":
        print("Isi parkiran:", parkiran)
    
    elif pilih == "4":
        break
```

```python
buku = []

while True:
    print("1. Tambah Buku")
    print("2. Ambil Buku Teratas")
    print("3. Lihat Buku Teratas")
    print("4. Tampilkan Semua Buku")
    print("5. Keluar")
    
    pilih = input("Pilih: ")
    
    if pilih == "1":
        judul = input("Judul buku: ")
        buku.append(judul)
    
    elif pilih == "2":
        if buku:
            print("Buku diambil:", buku.pop())
        else:
            print("Tidak ada buku")
    
    elif pilih == "3":
        if buku:
            print("Buku teratas:", buku[-1])
        else:
            print("Stack kosong")
    
    elif pilih == "4":
        print("Daftar buku:", buku)
    
    elif pilih == "5":
        break
```
