
# Tugas Struktur Data — Tree dan Parse Tree

Notebook ini berisi:

1. Implementasi Tree
2. Program Trace Parse Tree
3. Membaca Token
4. Menampilkan Current Node
5. Menampilkan Parse Tree


````python

# Class Node untuk Tree dan Parse Tree

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
````


## Fungsi Menampilkan Parse Tree


````python

def print_tree(node, level=0):
    if node is not None:
        print_tree(node.right, level + 1)
        print(' ' * 4 * level + '->', node.value)
        print_tree(node.left, level + 1)
````


## Input Ekspresi
Contoh ekspresi:
- a+b
- x*y
- p-q


````python

# Input ekspresi sederhana

expression = input("Masukkan ekspresi sederhana (contoh: a+b): ")

# Tokenisasi
tokens = list(expression)

print("\nToken yang dibaca:")
for token in tokens:
    print("Membaca token:", token)
````


## Membuat Parse Tree


````python

# Validasi ekspresi sederhana

if len(tokens) != 3:
    print("Format harus seperti: a+b")
else:
    left_operand = tokens[0]
    operator = tokens[1]
    right_operand = tokens[2]

    print("\nCurrent Node:", operator)

    # Membuat parse tree
    root = Node(operator)
    root.left = Node(left_operand)
    root.right = Node(right_operand)

    print("\nParse Tree:")
    print_tree(root)
````
