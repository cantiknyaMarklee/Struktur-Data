# Linked List – Unordered List dan Ordered List

Notebook ini berisi implementasi:
1. Unordered List
2. Ordered List
3. Operasi Search dan Insert angka 18



## Penjelasan Soal

### Unordered List
Aturan:
1. Jika angka yang dicari ada, maka angka 18 disisipkan setelah angka tersebut.
2. Jika angka tidak ada, maka angka 18 disisipkan di akhir linked list.

Contoh:
- Head → 7 → 18 → 6 → None
  menjadi:
- Head → 7 → 18 → 18 → 6 → None

Jika tidak ditemukan:
- Head → 7 → 10 → 6 → None
  menjadi:
- Head → 7 → 10 → 6 → 18 → None

---

### Ordered List
Aturan:
1. Linked list terurut.
2. Iterasi dilakukan hingga:
   - data baru lebih kecil dari node sekarang.
3. Jika data ditemukan:
   - sisipkan setelah node tersebut.
4. Jika data tidak ditemukan:
   - sisipkan sebelum node yang lebih besar.


```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class UnorderedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new_node

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" --> ")
            temp = temp.next

        print("None")

    def search_insert(self, target, value):
        temp = self.head

        while temp:
            # jika ditemukan
            if temp.data == target:
                new_node = Node(value)

                # sambungkan node baru ke next lama
                new_node.next = temp.next

                # node sekarang sambung ke node baru
                temp.next = new_node

                print(f"{target} ditemukan, {value} disisipkan setelahnya.")
                return

            temp = temp.next

        # jika tidak ditemukan -> insert di akhir
        self.append(value)
        print(f"{target} tidak ditemukan, {value} ditambahkan di akhir.")


# CONTOH 1
print("=== CONTOH 1 ===")
ulist1 = UnorderedList()

ulist1.append(7)
ulist1.append(18)
ulist1.append(6)

print("Sebelum Insert:")
ulist1.display()

ulist1.search_insert(18, 18)

print("Sesudah Insert:")
ulist1.display()


# CONTOH 2
print("\n=== CONTOH 2 ===")
ulist2 = UnorderedList()

ulist2.append(7)
ulist2.append(10)
ulist2.append(6)

print("Sebelum Insert:")
ulist2.display()

ulist2.search_insert(18, 18)

print("Sesudah Insert:")
ulist2.display()
```

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class OrderedList:
    def __init__(self):
        self.head = None

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" --> ")
            temp = temp.next

        print("None")

    def insert(self, value):
        new_node = Node(value)

        # jika head kosong
        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        prev = None

        while temp and temp.data < value:
            prev = temp
            temp = temp.next

        # sisipkan di awal
        if prev is None:
            new_node.next = self.head
            self.head = new_node

        else:
            prev.next = new_node
            new_node.next = temp

    def search_insert(self, target, value):
        temp = self.head

        while temp:
            # jika target ditemukan
            if temp.data == target:
                new_node = Node(value)

                new_node.next = temp.next
                temp.next = new_node

                print(f"{target} ditemukan, {value} disisipkan setelahnya.")
                return

            temp = temp.next

        # jika target tidak ditemukan
        self.insert(value)
        print(f"{target} tidak ditemukan, {value} disisipkan sesuai urutan.")


olist = OrderedList()

olist.insert(5)
olist.insert(10)
olist.insert(20)

print("Sebelum Insert:")
olist.display()

olist.search_insert(10, 18)

print("Sesudah Insert:")
olist.display()


print("\nContoh target tidak ditemukan")

olist2 = OrderedList()

olist2.insert(5)
olist2.insert(10)
olist2.insert(20)

print("Sebelum Insert:")
olist2.display()

olist2.search_insert(15, 18)

print("Sesudah Insert:")
olist2.display()
```


## Kesimpulan

### Unordered List
- Jika target ditemukan:
  - node baru disisipkan setelah target.
- Jika target tidak ditemukan:
  - node baru ditambahkan di akhir linked list.

### Ordered List
- Jika target ditemukan:
  - node baru disisipkan setelah target.
- Jika target tidak ditemukan:
  - node baru disisipkan sesuai urutan ascending.
